```python
import numpy as np
import scipy as sp
import scipy.special as spsp
import matplotlib.pyplot as plt
```

## Binomial a Normal (CLT)

```python
n = 100 # 30
p = 0.7 # para p muy pequenotodavia se tiene probabilidad para la normay deja de tener sentido, aqui necesitamos la Poisson
q = 1-p

# Creamos distribucion binomial
PXbinomial = []
for k in range(n+1):
    probX = spsp.comb(n,k) * p**k * q**(n-k)
    PXbinomial.append(probX)
    
# Creamos distribucion normal
mu = n*p
sigma = np.sqrt(n*p*q)

X = np.arange(0, n, 0.01)
PXnormal = (1/(sigma*np.sqrt(2*np.pi)))*np.exp(-((X-mu)**2)/(2*sigma**2))

# Plot
plt.plot(PXbinomial, label="binomial")
plt.plot(X, PXnormal, label="normal")
plt.legend()
plt.xlim(0,n)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# Set the random seed for reproducibility
# np.random.seed(42)

# Genera N=10000 random samples from a Poisson distribition with a mean lambda
lambda_value= 10
N = 10000
DataPopulation = np.random.poisson(lam=lambda_value, size=N)
```

```python
# Parameter for sampling
n = 100 
m = 1000

SampleMeans = np.zeros(m)
SampleSTDs = np.zeros(m)

for i in range(m):
    DataSample = np.random.choice( DataPopulation, size=n, replace=False)
    SampleMeans[i] = np.mean(DataSample)
    SampleSTDs[i] = np.std(DataPopulation)
# Calculate the population mean and SD
mean_population = np.mean(DataPopulation)
std_population = np.std(DataPopulation)
    
```

```python
std_error = std_population / np.sqrt(n)

x = np.linspace(min(SampleMeans), max(SampleMeans), 100)
normal_approximation = norm.pdf(x, loc=mean_population, scale=std_error)
```

```python
# plot the histogram of the SampleMeans
plt.figure(figsize=(10,6))
plt.hist(SampleMeans, bins=15, color='red', edgecolor='black', alpha=0.7, density=True)

plt.plot(x, normal_approximation, 'b-', label=f'Normal Approximation\n(mean={mean_population}, std_error={std_error})'  )

         
plt.xlabel('Sample Mean')
plt.ylabel('Density')
plt.title('Histogram of Sample Means with Normal Approximation')

plt.legend()
plt.show()
```

## Garant\'ias Estad\'isticas y Teorema del L\'imite Central

En la pr\'actica del Machine Learning, a menudo necesitamos evaluar el rendimiento de un modelo. Suponga que desea estimar la precisi\'on (accuracy) real $p$ de un nuevo clasificador utilizando un conjunto de prueba (test set) de tama\~no $n$. Queremos garantizar con un 95\% de confianza que la precisi\'on emp\'irica observada $\bar{X}_n$ no se desv\'ie de la verdadera precisi\'on $p$ por m\'as de un 2\% ($\varepsilon = 0.02$). Asuma el peor escenario de varianza para un ensayo de Bernoulli.

- Utilizando la **Desigualdad de Chebyshev**, determine el tama\~no de muestra $n$ estrictamente necesario para cumplir esta garant\'ia.
- Utilizando la aproximaci\'on asint\'otica del **Teorema del L\'imite Central (CLT)**, calcule el nuevo tama\~no de muestra $n$ requerido. (Puede asumir que $Z_{0.975} \approx 1.96$).
- **Simulaci\'on Computacional:** Suponga que el clasificador tiene una precisi\'on real y oculta de $p = 0.85$. Usando el $n$ que obtuvo en la parte (b), programe una simulaci\'on que genere 1000 conjuntos de prueba distintos (utilizando ensayos de Bernoulli). Calcule la precisi\'on emp\'irica para cada uno de los 1000 conjuntos y grafique un histograma con estos resultados. 
- Basado en los resultados de su simulaci\'on en (c), calcule la fracci\'on emp\'irica de los conjuntos de prueba cuya precisi\'on cay\'o dentro del margen de error permitido $[0.83, 0.87]$. ?`Corrobora esto la aproximaci\'on te\'orica del CLT? Explique brevemente.

```python
import numpy as np
import matplotlib.pyplot as plt

# Fijar semilla para reproducibilidad en la pauta
np.random.seed(42)

# --- PARTE (c) ---
p_real = 0.85
n_muestras = 2401  # Resultado obtenido en la parte (b)
num_experimentos = 1000

# Simulamos 1000 conjuntos de prueba usando ensayos de Bernoulli.
# Generamos una matriz de (1000, 2401) con valores entre 0 y 1.
# Si el valor es < p_real, es un acierto (1), si no, un error (0).
simulaciones = np.random.rand(num_experimentos, n_muestras) < p_real

# Calculamos la precisión (accuracy) empírica de cada uno de los 1000 conjuntos
precisiones_empiricas = simulaciones.mean(axis=1)

# Graficar el Histograma
plt.figure(figsize=(10, 6))
plt.hist(precisiones_empiricas, bins=30, color='royalblue', edgecolor='white', alpha=0.7)
plt.axvline(x=p_real, color='red', linestyle='--', linewidth=2, label=f'Precisión Real (p={p_real})')
plt.axvline(x=0.83, color='orange', linestyle=':', linewidth=2, label='Límite Inferior (0.83)')
plt.axvline(x=0.87, color='orange', linestyle=':', linewidth=2, label='Límite Superior (0.87)')

plt.title('Distribución de Precisiones Empíricas (1000 Test Sets)', fontsize=14)
plt.xlabel('Precisión Empírica de la Muestra', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()

# --- PARTE (d) ---
# Calcular empíricamente cuántas veces caímos en el margen [0.83, 0.87]
dentro_del_margen = np.sum((precisiones_empiricas >= 0.83) & (precisiones_empiricas <= 0.87))
fraccion_empirica = dentro_del_margen / num_experimentos

print(f"Fracción empírica dentro del margen [0.83, 0.87]: {fraccion_empirica * 100:.1f}%")
```
